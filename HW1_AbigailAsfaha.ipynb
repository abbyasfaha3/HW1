```python
O(1) is the constant time complexity; it is considered best-case/optimal.
O(log N) is the logarithm time complexity; it is considered very good/efficient.
O(N) is the linear time complexity; it is considered moderate/acceptable.
O(N^2) is the quadratic time complexity; it is considered worst-case/inefficient.
```

```python
A binary search initially finds the midpoint (element in the middle of the sorted array).
Then it compares that midpoint value to the target value (value you are searching for).
If the target value is bigger than the midpoint value, the first half of the array would be ignored, and the second half would be searched.
If the target value is smaller than the midpoint value, the second half of the array would be ignored, and the first half would be searched.
```
